```python
import nltk
from nltk import word_tokenize, bigrams
from collections import Counter
import matplotlib.pyplot as plt

# Certifique-se de ter baixado os recursos necessários
nltk.download('punkt')

# Texto de exemplo
poema = """
No meio do caminho tinha uma pedra
tinha uma pedra no meio do caminho
tinha uma pedra
no meio do caminho tinha uma pedra.
"""

# Tokenização (divisão do texto em palavras)
tokens = word_tokenize(poema.lower())

# Filtrar apenas palavras (removendo pontuação)
tokens = [t for t in tokens if t.isalpha()]

# Contagem de tokens
num_tokens = len(tokens)

# Geração de pares de palavras (bigrams)
pares = list(bigrams(tokens))
num_pares = len(pares)

# Contagem de frequência dos pares
contagem_pares = Counter(pares)

# Exibir resultados no console
print(f"Número total de tokens: {num_tokens}")
print(f"Número total de pares de palavras: {num_pares}")
print("\nPares de palavras mais comuns:")
for par, freq in contagem_pares.most_common(5):
    print(f"{par}: {freq}")

# Criar um gráfico
fig, ax = plt.subplots(1, 2, figsize=(10, 4))

# Gráfico 1: contagem geral
ax[0].bar(['Tokens', 'Pares'], [num_tokens, num_pares], color=['skyblue', 'lightgreen'])
ax[0].set_title('Contagem Geral')
ax[0].set_ylabel('Quantidade')

# Gráfico 2: pares mais comuns
pares_comuns = contagem_pares.most_common(5)
if pares_comuns:
    palavras = [' '.join(p) for p, _ in pares_comuns]
    frequencias = [f for _, f in pares_comuns]
    ax[1].barh(palavras, frequencias, color='salmon')
    ax[1].set_title('Pares de Palavras Mais Frequentes')
    ax[1].invert_yaxis()  # inverter para o par mais frequente ficar no topo

plt.tight_layout()
plt.show()
```
